**Univariate analysis**  : is the simplest form of statistical analysis that examines only one variable at a time.

______________________________________________________________________________

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("/content/train.csv")
df.head()
```

```python
df.info()
```

**Categorical Data**

Count Plot

```python
sns.countplot(x=df['Pclass'])
df['Pclass'].value_counts()
```

Pie Chart

```python
df["Pclass"].value_counts().plot(kind='pie', autopct='%.2f')     # Get percantage
# df["Pclass"].value_counts().plot(kind='pie')                   # without Percantage
```

**Numerical Data**

Histogram

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['Age'], bins=30)    # Aslo Use Bins
```

Displot

Smooth Curve Line Known as PDF (Probability Density Function)

```python
sns.displot(df['Age'], kde=True, bins=20)
```

```python
sns.boxplot(df['Age'])
```

```python
df['Age'].min()
```

```python
df['Age'].max()
```

```python
df['Age'].mean()
```

```python
titanic.head()
```

```python
sns.barplot(x=titanic['Pclass'], y=titanic['Age'], hue=titanic['Sex'])
```

______________

**2nd: ( Numeric to Categoricla )**

```python
sns.boxplot(x=titanic['Sex'], y=titanic['Age'], hue=titanic['Survived'])
```

________________

**DistPlot**

```python
sns.distplot(titanic[titanic['Survived']==0] ['Age'], hist=False, color='red')            # Survived
sns.distplot(titanic[titanic['Survived']==1] ['Age'], hist=False, color='green')          # Not Survived
```

________________________________

**3rd: (Categorical to Categorical)**

**HeatMap**

```python
titanic.head()
```

```python
pd.crosstab(titanic['Pclass'], titanic['Survived'])
```

```python
sns.heatmap(pd.crosstab(titanic['Pclass'], titanic['Survived']))
```
