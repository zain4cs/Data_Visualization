```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv("/content/train.csv")
df.head()
```

```python
df.info()
```

**Univariate analysis**  : is the simplest form of statistical analysis that examines only one variable at a time.

**Categorical Data**

Count Plot

```python
sns.countplot(x=df['Pclass'])
df['Pclass'].value_counts()
```

Pie Chart

```python
df["Pclass"].value_counts().plot(kind='pie', autopct='%.2f')     # Get percantage
# df["Pclass"].value_counts().plot(kind='pie')                   # without Percantage
```

**Numerical Data**

Histogram

```python
import matplotlib.pyplot as plt
```

```python
plt.hist(df['Age'], bins=30)    # Aslo Use Bins
```

Displot

Smooth Curve Line Known as PDF (Probability Density Function)

```python
sns.displot(df['Age'], kde=True, bins=20)
```
